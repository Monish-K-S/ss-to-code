```python
from pathlib import Path
from collections import Counter
import json

WEB_DIR = Path("/content/pix2code/datasets/extracted/web/all_data")

SPECIAL_TOKENS = ["<PAD>", "<START>", "<END>", "<UNK>"]

# Collect all DSL tokens
counter = Counter()

for gui_file in WEB_DIR.glob("*.gui"):
    text = gui_file.read_text().strip()

    tokens = (
        text
        .replace("{", " { ")
        .replace("}", " } ")
        .replace(",", " , ")
        .split()
    )

    counter.update(tokens)

# Build vocabulary
vocab_tokens = SPECIAL_TOKENS + sorted(counter.keys())

token_to_id = {
    token: idx
    for idx, token in enumerate(vocab_tokens)
}

id_to_token = {
    idx: token
    for token, idx in token_to_id.items()
}

print("Vocabulary size:", len(token_to_id))

for token, idx in token_to_id.items():
    print(f"{idx:2d} -> {token}")
```

```python
PAD_ID = token_to_id["<PAD>"]
START_ID = token_to_id["<START>"]
END_ID = token_to_id["<END>"]
UNK_ID = token_to_id["<UNK>"]


def tokenize_gui(gui_path):
    text = Path(gui_path).read_text().strip()

    tokens = (
        text
        .replace("{", " { ")
        .replace("}", " } ")
        .replace(",", " , ")
        .split()
    )

    return tokens


def encode_gui(gui_path):
    tokens = tokenize_gui(gui_path)

    token_ids = [START_ID]

    for token in tokens:
        token_ids.append(
            token_to_id.get(token, UNK_ID)
        )

    token_ids.append(END_ID)

    return token_ids


def decode_ids(token_ids):
    tokens = []

    for idx in token_ids:
        token = id_to_token[int(idx)]

        if token in ["<PAD>", "<START>"]:
            continue

        if token == "<END>":
            break

        tokens.append(token)

    return tokens
```

```python
import torch
from torch.utils.data import Dataset
from PIL import Image


MAX_SEQ_LEN = 120


class ScreenshotToHTMLDataset(Dataset):

    def __init__(self, samples, transform=None):
        self.samples = samples
        self.transform = transform

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):

        sample = self.samples[idx]

        # Load image
        image = Image.open(sample["image"]).convert("RGB")

        if self.transform:
            image = self.transform(image)

        # Encode DSL
        token_ids = encode_gui(sample["gui"])

        # Convert to tensor
        token_ids = torch.tensor(
            token_ids,
            dtype=torch.long
        )

        return image, token_ids
```

```python
from torch.nn.utils.rnn import pad_sequence


def collate_fn(batch):

    images, sequences = zip(*batch)

    images = torch.stack(images)

    sequences = pad_sequence(
        sequences,
        batch_first=True,
        padding_value=PAD_ID
    )

    return images, sequences
```

```python
import torch
import torch.nn as nn


class CNNEncoder(nn.Module):

    def __init__(self, feature_dim=256):
        super().__init__()

        self.features = nn.Sequential(

            # 224 → 112
            nn.Conv2d(3, 32, kernel_size=3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.MaxPool2d(2),

            # 112 → 56
            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.MaxPool2d(2),

            # 56 → 28
            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(),
            nn.MaxPool2d(2),

            # 28 → 1
            nn.AdaptiveAvgPool2d((1, 1))
        )

        self.fc = nn.Linear(128, feature_dim)

    def forward(self, x):

        x = self.features(x)

        x = x.view(x.size(0), -1)

        x = self.fc(x)

        return x
```

```python
class LSTMDecoder(nn.Module):

    def __init__(
        self,
        vocab_size,
        feature_dim=256,
        embedding_dim=128,
        hidden_dim=256
    ):
        super().__init__()

        self.embedding = nn.Embedding(
            vocab_size,
            embedding_dim,
            padding_idx=PAD_ID
        )

        self.image_to_hidden = nn.Linear(
            feature_dim,
            hidden_dim
        )

        self.image_to_cell = nn.Linear(
            feature_dim,
            hidden_dim
        )

        self.lstm = nn.LSTM(
            input_size=embedding_dim,
            hidden_size=hidden_dim,
            batch_first=True
        )

        self.fc = nn.Linear(
            hidden_dim,
            vocab_size
        )

    def forward(self, image_features, captions):

        # Convert CNN feature to initial LSTM states
        hidden = self.image_to_hidden(
            image_features
        ).unsqueeze(0)

        cell = self.image_to_cell(
            image_features
        ).unsqueeze(0)

        # Token embeddings
        embeddings = self.embedding(captions)

        # LSTM
        outputs, _ = self.lstm(
            embeddings,
            (hidden, cell)
        )

        # Vocabulary prediction
        outputs = self.fc(outputs)

        return outputs
```

```python
class ScreenshotToCodeModel(nn.Module):

    def __init__(self, vocab_size):
        super().__init__()

        self.encoder = CNNEncoder(
            feature_dim=256
        )

        self.decoder = LSTMDecoder(
            vocab_size=vocab_size,
            feature_dim=256,
            embedding_dim=128,
            hidden_dim=256
        )

    def forward(self, images, captions):

        features = self.encoder(images)

        outputs = self.decoder(
            features,
            captions
        )

        return outputs
```

```python
from tqdm.auto import tqdm


def train_one_epoch(model, loader, optimizer, criterion, device):

    model.train()

    total_loss = 0.0

    progress = tqdm(
        loader,
        desc="Training",
        leave=False
    )

    for images, sequences in progress:

        images = images.to(
            device,
            non_blocking=True
        )

        sequences = sequences.to(
            device,
            non_blocking=True
        )

        # Teacher forcing
        decoder_input = sequences[:, :-1]
        targets = sequences[:, 1:]

        # Forward pass
        outputs = model(
            images,
            decoder_input
        )

        # [batch, seq, vocab]
        # → [batch * seq, vocab]
        outputs = outputs.reshape(
            -1,
            outputs.size(-1)
        )

        targets = targets.reshape(-1)

        loss = criterion(
            outputs,
            targets
        )

        optimizer.zero_grad()

        loss.backward()

        # Prevent exploding gradients
        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            max_norm=5.0
        )

        optimizer.step()

        total_loss += loss.item()

        progress.set_postfix(
            loss=f"{loss.item():.4f}"
        )

    return total_loss / len(loader)
```

```python
def validate(model, loader, criterion, device):

    model.eval()

    total_loss = 0.0

    with torch.no_grad():

        for images, sequences in loader:

            images = images.to(
                device,
                non_blocking=True
            )

            sequences = sequences.to(
                device,
                non_blocking=True
            )

            decoder_input = sequences[:, :-1]
            targets = sequences[:, 1:]

            outputs = model(
                images,
                decoder_input
            )

            outputs = outputs.reshape(
                -1,
                outputs.size(-1)
            )

            targets = targets.reshape(-1)

            loss = criterion(
                outputs,
                targets
            )

            total_loss += loss.item()

    return total_loss / len(loader)
```

```python
EPOCHS = 5

history = {
    "train_loss": [],
    "val_loss": []
}

for epoch in range(EPOCHS):

    train_loss = train_one_epoch(
        model,
        train_loader,
        optimizer,
        criterion,
        device
    )

    val_loss = validate(
        model,
        val_loader,
        criterion,
        device
    )

    history["train_loss"].append(train_loss)
    history["val_loss"].append(val_loss)

    print(
        f"Epoch {epoch + 1}/{EPOCHS} | "
        f"Train Loss: {train_loss:.4f} | "
        f"Val Loss: {val_loss:.4f}"
    )
```

```python
def generate_sequence(
    model,
    image,
    max_length=120
):
    model.eval()

    with torch.no_grad():

        # Add batch dimension
        image = image.unsqueeze(0).to(device)

        # CNN encoder
        features = model.encoder(image)

        # Initial LSTM states
        hidden = model.decoder.image_to_hidden(
            features
        ).unsqueeze(0)

        cell = model.decoder.image_to_cell(
            features
        ).unsqueeze(0)

        # Start token
        current_token = torch.tensor(
            [[START_ID]],
            dtype=torch.long,
            device=device
        )

        generated = []

        for _ in range(max_length):

            embedding = model.decoder.embedding(
                current_token
            )

            output, (hidden, cell) = model.decoder.lstm(
                embedding,
                (hidden, cell)
            )

            logits = model.decoder.fc(output)

            # Greedy decoding
            next_token = logits.argmax(
                dim=-1
            )

            token_id = next_token.item()

            if token_id == END_ID:
                break

            generated.append(token_id)

            current_token = next_token

        return generated
```

```python
def levenshtein_distance(a, b):
    """
    Compute Levenshtein edit distance between two token sequences.
    """
    m = len(a)
    n = len(b)

    dp = [[0] * (n + 1) for _ in range(m + 1)]

    for i in range(m + 1):
        dp[i][0] = i

    for j in range(n + 1):
        dp[0][j] = j

    for i in range(1, m + 1):
        for j in range(1, n + 1):

            cost = 0 if a[i - 1] == b[j - 1] else 1

            dp[i][j] = min(
                dp[i - 1][j] + 1,
                dp[i][j - 1] + 1,
                dp[i - 1][j - 1] + cost
            )

    return dp[m][n]


def sequence_similarity(predicted, target):
    """
    Returns similarity in the range [0, 1].
    """
    if len(predicted) == 0 and len(target) == 0:
        return 1.0

    distance = levenshtein_distance(
        predicted,
        target
    )

    return 1 - (
        distance / max(len(predicted), len(target))
    )
```

```python
import time
import numpy as np


def evaluate_model(model, dataset):

    model.eval()

    total_correct = 0
    total_target_tokens = 0

    exact_matches = 0
    similarities = []

    inference_times = []

    examples = []

    for idx in range(len(dataset)):

        image, target_sequence = dataset[idx]

        target_tokens = decode_ids(
            target_sequence.tolist()
        )

        # Generate
        start_time = time.perf_counter()

        generated_ids = generate_sequence(
            model,
            image,
            max_length=120
        )

        elapsed = time.perf_counter() - start_time

        inference_times.append(elapsed)

        generated_tokens = decode_ids(
            generated_ids
        )

        # ------------------------------------------------
        # Token accuracy
        # ------------------------------------------------

        correct = sum(
            pred == target
            for pred, target
            in zip(
                generated_tokens,
                target_tokens
            )
        )

        total_correct += correct
        total_target_tokens += len(target_tokens)

        # ------------------------------------------------
        # Exact sequence match
        # ------------------------------------------------

        if generated_tokens == target_tokens:
            exact_matches += 1

        # ------------------------------------------------
        # Sequence similarity
        # ------------------------------------------------

        similarities.append(
            sequence_similarity(
                generated_tokens,
                target_tokens
            )
        )

        # Save a few examples
        if len(examples) < 5:
            examples.append({
                "index": idx,
                "target": target_tokens,
                "generated": generated_tokens
            })

    token_accuracy = (
        total_correct / total_target_tokens
    )

    exact_match = (
        exact_matches / len(dataset)
    )

    avg_similarity = np.mean(
        similarities
    )

    avg_inference_time = np.mean(
        inference_times
    )

    return {
        "token_accuracy": token_accuracy,
        "exact_match": exact_match,
        "sequence_similarity": avg_similarity,
        "avg_inference_time": avg_inference_time,
        "examples": examples
    }
```

```python
import torch
import torch.nn as nn
from torchvision.models import resnet18, ResNet18_Weights


class ResNet18Encoder(nn.Module):

    def __init__(self, feature_dim=256):
        super().__init__()

        # Load ImageNet-pretrained ResNet18
        self.resnet = resnet18(
            weights=ResNet18_Weights.DEFAULT
        )

        # Remove original classification layer
        self.resnet.fc = nn.Identity()

        # ResNet18 outputs 512 features
        self.fc = nn.Linear(
            512,
            feature_dim
        )

    def forward(self, x):

        x = self.resnet(x)

        x = self.fc(x)

        return x
```

```python
class ResNetScreenshotToCodeModel(nn.Module):

    def __init__(self, vocab_size):

        super().__init__()

        self.encoder = ResNet18Encoder(
            feature_dim=256
        )

        self.decoder = LSTMDecoder(
            vocab_size=vocab_size,
            feature_dim=256,
            embedding_dim=128,
            hidden_dim=256
        )

    def forward(self, images, captions):

        features = self.encoder(images)

        outputs = self.decoder(
            features,
            captions
        )

        return outputs
```
